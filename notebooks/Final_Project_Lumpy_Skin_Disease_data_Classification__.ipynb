```python
# Select only the numeric columns
numeric_df = df.select_dtypes(include=[float, int])

# Removing the outliers
for col in numeric_df:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    upper_limit = q3 + (1.5 * iqr)
    lower_limit = q1 - (1.5 * iqr)
    new_df2 = df.loc[(df[col] <= upper_limit) & (df[col] >= lower_limit)]
print('before removing outliers: ', len(df))
print('after removing outliers: ', len(new_df2))
print('outliers:', len(df)-len(new_df2))
```

```python
#Import Libraries
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    roc_curve
)

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Import Models
from sklearn.linear_model import LogisticRegression

from sklearn.tree import DecisionTreeClassifier

from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier,
    VotingClassifier,
    StackingClassifier
)

from sklearn.naive_bayes import GaussianNB

from sklearn.neighbors import KNeighborsClassifier

from sklearn.svm import SVC

from sklearn.neural_network import MLPClassifier

from xgboost import XGBClassifier

from lightgbm import LGBMClassifier

from catboost import CatBoostClassifier

from imblearn.pipeline import Pipeline

from imblearn.over_sampling import SMOTE
```

```python
#Build Models
models = {

    "Logistic Regression":
        LogisticRegression(max_iter=1000),

    "Decision Tree":
        DecisionTreeClassifier(random_state=42),

    "Random Forest":
        RandomForestClassifier(
            n_estimators=300,
            random_state=42
        ),

    "Gaussian NB":
        GaussianNB(),

    "KNN":
        KNeighborsClassifier(n_neighbors=5),

    "SVM":
        SVC(
            probability=True,
            random_state=42
        ),

    "Gradient Boosting":
        GradientBoostingClassifier(random_state=42),

    "XGBoost":
        XGBClassifier(
            eval_metric="logloss",
            random_state=42
        ),

    "LightGBM":
        LGBMClassifier(random_state=42),

    "CatBoost":
        CatBoostClassifier(
            verbose=0,
            random_state=42
        ),

    "Neural Network":
        MLPClassifier(
            hidden_layer_sizes=(128,64),
            max_iter=500,
            random_state=42
        )
}
```

```python
#Random Forest + SMOTE
rf_smote = Pipeline([

    ("smote", SMOTE(random_state=42)),

    ("rf", RandomForestClassifier(
        n_estimators=300,
        random_state=42
    ))

])

models["Random Forest + SMOTE"] = rf_smote
```

```python
#Voting Ensemble
voting = VotingClassifier(

    estimators=[

        ("rf", RandomForestClassifier(random_state=42)),

        ("xgb", XGBClassifier(
            eval_metric="logloss",
            random_state=42
        )),

        ("cat", CatBoostClassifier(
            verbose=0,
            random_state=42
        ))

    ],

    voting="soft"

)

models["Voting Ensemble"] = voting
```

```python
#Stacking Ensemble
stacking = StackingClassifier(

    estimators=[

        ("rf", RandomForestClassifier(random_state=42)),

        ("gb", GradientBoostingClassifier(random_state=42)),

        ("xgb", XGBClassifier(
            eval_metric="logloss",
            random_state=42
        ))

    ],

    final_estimator=LogisticRegression()

)

models["Stacking Ensemble"] = stacking
```

```python
#Train All Models
results = []

trained_models = {}

for name, model in models.items():

    print("="*70)
    print(f"Training {name}")

    model.fit(X_train, y_train)

    trained_models[name] = model

    y_pred = model.predict(X_test)

    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:,1]
    else:
        y_prob = None

    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(y_test, y_pred)

    recall = recall_score(y_test, y_pred)

    f1 = f1_score(y_test, y_pred)

    if y_prob is not None:
        roc_auc = roc_auc_score(y_test, y_prob)
    else:
        roc_auc = np.nan

    results.append({

        "Model":name,

        "Accuracy":accuracy,

        "Precision":precision,

        "Recall":recall,

        "F1":f1,

        "ROC-AUC":roc_auc

    })
```

#8.Model Evaluation

```python
#Results Table
results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="F1",
    ascending=False
)

results_df.reset_index(drop=True,inplace=True)

results_df
```

```python
#Best Model
best_model_name = results_df.iloc[0]["Model"]

best_model = trained_models[best_model_name]

print("Best Model :", best_model_name)
```

```python
#Classification Report
y_pred = best_model.predict(X_test)

print(classification_report(y_test,y_pred))
```

```python
#Confusion Matrix
cm = confusion_matrix(y_test,y_pred)

plt.figure(figsize=(6,5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title(best_model_name)

plt.xlabel("Predicted")

plt.ylabel("Actual")

plt.show()
```

```python
#ROC Curve
if hasattr(best_model,"predict_proba"):

    y_prob = best_model.predict_proba(X_test)[:,1]

    fpr,tpr,_ = roc_curve(y_test,y_prob)

    plt.figure(figsize=(7,5))

    plt.plot(
        fpr,
        tpr,
        label=f"AUC = {roc_auc_score(y_test,y_prob):.4f}"
    )

    plt.plot([0,1],[0,1],"--")

    plt.xlabel("False Positive Rate")

    plt.ylabel("True Positive Rate")

    plt.title(best_model_name)

    plt.legend()

    plt.show()
```

```python
#Accuracy Comparison
plt.figure(figsize=(12,6))

sns.barplot(
    data=results_df,
    x="Accuracy",
    y="Model"
)

plt.title("Model Accuracy Comparison")

plt.show()
```

# 9.Hyperparameter Tuning

### RandomizedSearchCV for XGBoost


```python
from sklearn.model_selection import RandomizedSearchCV
from xgboost import XGBClassifier
import joblib
import pickle

# Define the parameter search space
params = {

    "n_estimators": [100, 200, 300, 500],

    "max_depth": [3, 5, 7, 9],

    "learning_rate": [0.01, 0.05, 0.1],

    "subsample": [0.7, 0.8, 1.0],

    "colsample_bytree": [0.7, 0.8, 1.0]

}

# Create the base XGBoost model
xgb = XGBClassifier(
    eval_metric="logloss",
    random_state=42
)

# Create the Randomized Search object
random_search = RandomizedSearchCV(

    estimator=xgb,

    param_distributions=params,

    n_iter=10,

    scoring="f1",

    cv=5,

    random_state=42,

    n_jobs=-1,

    verbose=2

)

# Train the model using Randomized Search
random_search.fit(X_train, y_train)

# Get the best model
best_xgb = random_search.best_estimator_

# Print the best hyperparameters
print("=" * 60)
print("Best Hyperparameters")
print(random_search.best_params_)

print("\nBest Cross Validation F1 Score")
print(random_search.best_score_)

# Evaluate the Best Model

y_pred = best_xgb.predict(X_test)

print("\nClassification Report")
print(classification_report(y_test, y_pred))

# Save the Best Model

joblib.dump(best_xgb, "Best_XGBoost_Model.joblib")

with open("Best_XGBoost_Model.pkl", "wb") as file:
    pickle.dump(best_xgb, file)

print("\nBest model saved successfully!")

# Save Feature Names

pickle.dump(
    list(X.columns),
    open("feature_names.pkl", "wb")
)

print("Feature names saved successfully!")
```

#10.Cross Validation

```python
from sklearn.model_selection import KFold, cross_val_score
import numpy as np

# K-Fold Cross Validation

kfold = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

kfold_scores = cross_val_score(
    estimator=best_xgb,
    X=X_train,
    y=y_train,
    cv=kfold,
    scoring="f1",
    n_jobs=-1
)

print("=" * 60)
print("K-Fold Cross Validation Results")
print("=" * 60)

print("Fold Scores:")
print(kfold_scores)

print("\nMean F1 Score:")
print(np.mean(kfold_scores))

print("\nStandard Deviation:")
print(np.std(kfold_scores))
```

```python
from sklearn.model_selection import StratifiedKFold

# Stratified K-Fold Cross Validation

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

stratified_scores = cross_val_score(
    estimator=best_xgb,
    X=X_train,
    y=y_train,
    cv=skf,
    scoring="f1",
    n_jobs=-1
)

print("=" * 60)
print("Stratified K-Fold Cross Validation Results")
print("=" * 60)

print("Fold Scores:")
print(stratified_scores)

print("\nMean F1 Score:")
print(np.mean(stratified_scores))

print("\nStandard Deviation:")
print(np.std(stratified_scores))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.plot(
    range(1, 6),
    stratified_scores,
    marker="o",
    linewidth=2
)

plt.title("Stratified K-Fold F1 Scores")
plt.xlabel("Fold")
plt.ylabel("F1 Score")
plt.grid(True)

plt.show()
```

```python
import gradio as gr
import pandas as pd
import joblib

# Load Trained Model

model = joblib.load("Best_Stacking_Model.joblib")

# Prediction Function

def predict_lumpy(
    longitude,
    latitude,
    cloud_cover,
    diurnal_temp_range,
    frost_days,
    potential_evapotranspiration,
    precipitation,
    min_temperature,
    mean_temperature,
    max_temperature,
    vapor_pressure,
    wet_days,
    elevation,
    land_cover,
    cattle_density,
    buffalo_density
):

    data = pd.DataFrame([[

        longitude,
        latitude,
        cloud_cover,
        diurnal_temp_range,
        frost_days,
        potential_evapotranspiration,
        precipitation,
        min_temperature,
        mean_temperature,
        max_temperature,
        vapor_pressure,
        wet_days,
        elevation,
        land_cover,
        cattle_density,
        buffalo_density

    ]],

    columns=[

        "x",
        "y",
        "cld",
        "dtr",
        "frs",
        "pet",
        "pre",
        "tmn",
        "tmp",
        "tmx",
        "vap",
        "wet",
        "elevation",
        "dominant_land_cover",
        "X5_Ct_2010_Da",
        "X5_Bf_2010_Da"

    ])

    prediction = model.predict(data)[0]

    if prediction == 1:
        return "🐄 Lumpy Skin Disease Detected"

    return "✅ Healthy Animal"

# Build Interface

demo = gr.Interface(

    fn=predict_lumpy,

    inputs=[

        gr.Number(label="Longitude (x)"),
        gr.Number(label="Latitude (y)"),
        gr.Number(label="Cloud Cover"),
        gr.Number(label="Diurnal Temperature Range"),
        gr.Number(label="Frost Days"),
        gr.Number(label="Potential Evapotranspiration"),
        gr.Number(label="Precipitation"),
        gr.Number(label="Minimum Temperature"),
        gr.Number(label="Mean Temperature"),
        gr.Number(label="Maximum Temperature"),
        gr.Number(label="Vapor Pressure"),
        gr.Number(label="Wet Days"),
        gr.Number(label="Elevation"),
        gr.Number(label="Dominant Land Cover"),
        gr.Number(label="Cattle Density"),
        gr.Number(label="Buffalo Density")

    ],

    outputs=gr.Textbox(label="Prediction"),

    title="🐄 Lumpy Skin Disease Detection System",

    description="""
Enter the environmental and geographical data, then click Predict.
The trained Stacking Ensemble model will classify the sample as Healthy or Diseased.
""",

    theme="soft"

)

demo.launch(share=True)
```
